```python
def generate_key_matrix(key):
    key = key.lower().replace('j', 'i')
    seen = set()
    key_unique = ''
    for char in key:
        if char.isalpha() and char not in seen:
            seen.add(char)
            key_unique += char

    # Fill remaining letters (excluding 'j')
    for char in 'abcdefghiklmnopqrstuvwxyz':
        if char not in seen:
            key_unique += char

    # Create 5x5 matrix
    matrix = [[None] * 5 for _ in range(5)]
    idx = 0
    for row in range(5):
        for col in range(5):
            matrix[row][col] = key_unique[idx]
            idx += 1
    return matrix


def encrypt(plaintext, key_matrix):
    plaintext = plaintext.replace('j', 'i').lower().replace(" ", "")
    ciphertext = ''
    i = 0
    while i < len(plaintext):
        a = plaintext[i]
        b = plaintext[i + 1] if i + 1 < len(plaintext) else 'x'
        if a == b:
            b = 'x'
            i += 1
        else:
            i += 2

        row1 = col1 = row2 = col2 = 0
        for r in range(5):
            for c in range(5):
                if key_matrix[r][c] == a:
                    row1, col1 = r, c
                if key_matrix[r][c] == b:
                    row2, col2 = r, c

        if row1 == row2:
            ciphertext += key_matrix[row1][(col1 + 1) % 5] + key_matrix[row2][(col2 + 1) % 5]
        elif col1 == col2:
            ciphertext += key_matrix[(row1 + 1) % 5][col1] + key_matrix[(row2 + 1) % 5][col2]
        else:
            ciphertext += key_matrix[row1][col2] + key_matrix[row2][col1]
    return ciphertext.upper()

def decrypt(ciphertext, key_matrix):
    plaintext = ''
    i = 0
    while i < len(ciphertext):
        a = ciphertext[i].lower()
        b = ciphertext[i + 1].lower()
        i += 2

        row1 = col1 = row2 = col2 = 0
        for r in range(5):
            for c in range(5):
                if key_matrix[r][c] == a:
                    row1, col1 = r, c
                if key_matrix[r][c] == b:
                    row2, col2 = r, c

        if row1 == row2:
            plaintext += key_matrix[row1][(col1 - 1) % 5] + key_matrix[row2][(col2 - 1) % 5]
        elif col1 == col2:
            plaintext += key_matrix[(row1 - 1) % 5][col1] + key_matrix[(row2 - 1) % 5][col2]
        else:
            plaintext += key_matrix[row1][col2] + key_matrix[row2][col1]

    return plaintext.replace('i', 'j')

# ---------- MAIN INTERACTION ----------
if __name__ == "__main__":
    print("Playfair Cipher Encryption/Decryption")
    key = input("Enter the key: ")
    mode = input("Choose mode (encrypt/decrypt): ").strip().lower()

    key_matrix = generate_key_matrix(key)

    print("\nKey Matrix:")
    for row in key_matrix:
        print(" ".join(row))

    if mode == "encrypt":
        plaintext = input("Enter plaintext: ")
        ciphertext = encrypt(plaintext, key_matrix)
        print("Encrypted ciphertext:", ciphertext)

    elif mode == "decrypt":
        ciphertext = input("Enter ciphertext: ")
        decrypted_text = decrypt(ciphertext, key_matrix)
        print("Decrypted plaintext:", decrypted_text.upper())

    else:
        print("Invalid mode selected.")
```
